```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches # Import patches for drawing rectangles
from ipywidgets import interact, FloatSlider
from IPython.display import display

# Funcion auxiliar que calcula el error cuadratico medio (MSE)
def RMS(actual, prediccion):
    # Nos aseguramos que ambos datos sean de la misma longitud
    actual_arr = np.array(actual)
    prediccion_arr = np.array(prediccion)
    return np.sum((actual_arr - prediccion_arr)**2)/len(prediccion_arr)

# Datos para ajustar (CAMBIA ESTOS VALORES)
x_data = np.array([0,1,2,3,4,5])
y_data = np.array([0,1.3,5,6,4,5.4])

# Ecuacion parametrica de la recta aplicada a un conjunto de datos
def aprox(sesgo, peso, puntos):
    x_approx_vals = np.linspace(0, puntos - 1, puntos)
    y_approx_vals = sesgo + peso * x_approx_vals
    return x_approx_vals, y_approx_vals

# Funcion para graficar el ejercicio
def tanteo(sesgo, peso):
    # --- Primera grafica y datos ---
    puntos = len(x_data)
    x_approx, y_approx = aprox(sesgo, peso, puntos)

    plt.figure(figsize=(14, 6))
    ax1 = plt.subplot(1, 2, 1) # Se genera una subgrafica para las areas de los cuadrados
    plt.scatter(x_data, y_data, label='Datos Originales')
    plt.plot(x_approx, y_approx, color='red', linestyle='-', label=f'Aproximacion: y = {sesgo:.2f} + {peso:.2f}x')
    plt.title(f'y = sesgo + peso x') # Intentar agregar Latex
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.grid(True)

    # Se agregan las areas de los cuadrados a la primer grafica
    for i in range(puntos):
        x_val = x_data[i]
        y_actual = y_data[i]
        y_predicted = y_approx[i] # Prediccion del valor de y

        error = y_actual - y_predicted
        abs_error = abs(error)

        # Se pinta una linea en cada punto que representa el residual
        ax1.plot([x_val, x_val], [y_predicted, y_actual], color='gray', linestyle='--', linewidth=1)

        # Pinta el area del cuadrado que representa el error
        if abs_error > 0.01: # Solo se grafica si el error es mayor de 0.01
            # Se grafica el cuadrado
            rect = patches.Rectangle((x_val - abs_error/2, y_predicted if error >= 0 else y_predicted - abs_error),
                                     abs_error, abs_error,
                                     linewidth=1, edgecolor='blue', facecolor='lightblue', alpha=0.3)
            ax1.add_patch(rect)

    # --- Segunda grafica: SME (error cuadratico medio) ---
    betas_range = np.linspace(-0.4, 3, 100) # Rango para que se muestre el error

    # Se calcula la suma de los errores al cuadrado para un peso y sesgo dado (parabola)
    sumas_squared_errors = []
    for b_val in betas_range:
        _, approx_y_for_b = aprox(sesgo, b_val, puntos)
        sumas_squared_errors.append(RMS(y_data, approx_y_for_b))

    # Se calcula el valor actual del SME para mostrarlo en la seguda grafica (punto del error en la parabola)
    _, current_y_approx_fixed_alfa = aprox(sesgo, peso, puntos)
    current_error_point = RMS(y_data, current_y_approx_fixed_alfa)

    plt.subplot(1, 2, 2) # Las graficas se muestran en un renglon de 2 columnas (1,2)
    plt.plot(betas_range, sumas_squared_errors, color='red', linestyle='-', label='Suma de los errores cuadrados')
    plt.scatter(peso, current_error_point, color='black', marker='o', s=100,
                label=f'Error Actual={current_error_point:.2f}') # Intentar agregar Latex
    plt.title(f'Error cuadratico') # Intentar agregar Latex
    plt.ylabel('Suma de los errores cuadrados')
    plt.legend()
    plt.grid(True)

    ax = plt.gca() # Obtiene el control de la subgráfica actual
    ax.spines['left'].set_position('zero')    # Mueve el eje Y al cero de X
    ax.spines['bottom'].set_position('zero')  # Mueve el eje X al cero de Y
    ax.spines['right'].set_color('none')      # Oculta el borde derecho
    ax.spines['top'].set_color('none')         # Oculta el borde superior
    plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
    plt.show()

# Se genera el slider mediante un widget que hace uso de la funcion tanteo
interact(tanteo,
         sesgo=FloatSlider(min=-1.0, max=3.0, step=0.1, value=0.0, description='Sesgo'),
         peso=FloatSlider(min=-1.0, max=3.0, step=0.1, value=0.0, description='Peso'))
```
