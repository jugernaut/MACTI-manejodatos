### El Concepto de Límite

En cálculo, el **límite de una función** es un concepto fundamental que describe el comportamiento de una función a medida que su variable independiente se "acerca" a un cierto valor. No se trata del valor de la función *en* ese punto, sino del valor al que la función *tiende* a medida que la entrada se aproxima.

Formalmente, decimos que el límite de una función $f(x)$ cuando $x$ se acerca a $c$ es $L$, lo cual se escribe como:

$$\lim_{x \to c} f(x) = L$$

Esto significa que podemos hacer que los valores de $f(x)$ estén tan cerca de $L$ como queramos, tomando valores de $x$ suficientemente cercanos a $c$ (pero no necesariamente iguales a $c$).

**¿Por qué es importante el concepto de límite?**

1.  **Continuidad:** Una función es continua en un punto si el límite de la función en ese punto existe y es igual al valor de la función en ese punto.
2.  **Derivada:** Como se vio anteriormente, la derivada se define mediante un límite. Es la base para entender cómo cambia una función.
3.  **Integrales:** Las integrales, que calculan áreas bajo curvas, también se definen mediante límites (sumas de Riemann).
4.  **Comportamiento asintótico:** Los límites nos ayudan a comprender el comportamiento de las funciones cuando $x$ tiende al infinito o a valores donde la función no está definida (asíntotas).

### Definición de la Derivada por Límites

La derivada de una función $f(x)$ en un punto $x$ se define como:

$$f'(x) = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}$$

Esta definición representa la pendiente de la línea tangente a la curva $f(x)$ en el punto $x$.

### La Derivada como Razón de Cambio y para Encontrar Máximos/Mínimos

La derivada de una función, $f'(x)$, no es solo la pendiente de la recta tangente en un punto, sino que también representa la **razón instantánea de cambio** de la función con respecto a su variable independiente $x$.

Imaginemos una gráfica de una función $f(x)$ que representa, por ejemplo, la distancia recorrida por un objeto en función del tiempo. La derivada $f'(x)$ en un instante dado nos dirá la velocidad instantánea del objeto en ese momento, es decir, cuán rápido está cambiando su posición. Si la derivada es positiva, la función está creciendo; si es negativa, la función está decreciendo; y si es cero, la función ha alcanzado un punto donde no está cambiando, lo que a menudo indica un máximo o un mínimo local.

**¿Por qué la derivada nos ayuda a encontrar máximos y mínimos?**

Los **máximos** y **mínimos** locales de una función ocurren en puntos donde la pendiente de la recta tangente es horizontal, es decir, donde la razón de cambio instantánea es cero. En estos puntos, la función deja de crecer para empezar a decrecer (un máximo) o deja de decrecer para empezar a crecer (un mínimo).

Para encontrar estos puntos:
1. Calculamos la primera derivada de la función, $f'(x)$.
2. Igualamos $f'(x)$ a cero y resolvemos para $x$. Los valores de $x$ obtenidos son los **puntos críticos**.
3. Para determinar si un punto crítico es un máximo, un mínimo o un punto de inflexión, podemos usar el criterio de la segunda derivada o analizar el signo de la primera derivada alrededor del punto.

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, FloatSlider
from IPython.display import display

# Define la función de ejemplo
def funcion_limite(x):
  return x**2

# Define la derivada analítica para la función de ejemplo (para la tangente)
def derivada_analitica_limite(x):
  return 2 * x

# Función para graficar de forma interactiva la aproximación por límites
def plot_limite_derivada(x0, h):
  x_valores = np.linspace(x0 - 2, x0 + 2, 400)
  y_valores = funcion_limite(x_valores)

  # Puntos para la recta secante
  x1 = x0
  y1 = funcion_limite(x1)
  x2 = x0 + h
  y2 = funcion_limite(x2)

  # Pendiente de la recta secante
  if h != 0:
    pendiente_secante = (y2 - y1) / h
  else:
    pendiente_secante = derivada_analitica_limite(x0) # Si h es 0, usamos la derivada real

  # Recta secante: y = m_secante * (x - x1) + y1
  x_secante = np.linspace(min(x1, x2) - 0.5, max(x1, x2) + 0.5, 100)
  y_secante = pendiente_secante * (x_secante - x1) + y1

  # Pendiente de la recta tangente (derivada analítica)
  pendiente_tangente = derivada_analitica_limite(x0)
  # Recta tangente: y = m_tangente * (x - x0) + f(x0)
  y_tangente = pendiente_tangente * (x_secante - x0) + funcion_limite(x0)

  plt.figure(figsize=(10, 6))
  plt.plot(x_valores, y_valores, label='f(x) = x²', color='blue')

  plt.scatter([x1, x2], [y1, y2], color='red', zorder=5, label=f'Puntos Secantes (h={h:.2f})')
  plt.plot(x_secante, y_secante, color='red', linestyle='--', label=f'Recta Secante (m={pendiente_secante:.2f})')

  plt.plot(x_secante, y_tangente, color='green', linestyle='-', label=f'Recta Tangente (m={pendiente_tangente:.2f})')

  plt.title(f'Concepto de Derivada usando Límites en x={x0:.2f}')
  plt.xlabel('x')
  plt.ylabel('f(x)')
  plt.grid(True)
  plt.axhline(0, color='gray', linewidth=0.8, linestyle=':')
  plt.axvline(0, color='gray', linewidth=0.8, linestyle=':')
  plt.legend()
  plt.ylim([min(y_valores) - 1, max(y_valores) + 1])
  plt.xlim([x0 - 2.5, x0 + 2.5]) # Ajusta el rango x para centrar en x0
  plt.show()

# Crear sliders interactivos
slider_x0 = FloatSlider(
    min=-2.0, max=2.0, step=0.1, value=0.5,
    description='Punto x0:'
)

slider_h = FloatSlider(
    min=0.001, max=1.5, step=0.01, value=1.0,
    description='Valor de h:'
)

print("Ajusta el punto 'x0' y el valor de 'h' para ver la recta secante acercarse a la tangente:")
interact(plot_limite_derivada, x0=slider_x0, h=slider_h);
```

### ¿Cuándo la Pendiente es Positiva?

La pendiente de una función en un punto dado (que es lo mismo que el valor de la derivada en ese punto) se considera **positiva** cuando la función está **aumentando** o **creciendo** en ese intervalo.

Visualmente, esto significa que si nos movemos de izquierda a derecha a lo largo del eje x:

*   La gráfica de la función **asciende**.
*   La línea tangente a la curva en ese punto se inclina **hacia arriba** de izquierda a derecha.

**En términos matemáticos:**

Si $f'(x) > 0$ (la derivada es mayor que cero) en un intervalo $(a, b)$, entonces la función $f(x)$ es **creciente** en ese intervalo. Esto implica que para cualquier par de puntos $x_1$ y $x_2$ en el intervalo, si $x_1 < x_2$, entonces $f(x_1) < f(x_2)$.

**Ejemplo:**

Considera la función $f(x) = x^2$. Su derivada es $f'(x) = 2x$.

*   Para $x > 0$ (es decir, en el intervalo $(0, \infty)$), la derivada $f'(x) = 2x$ es positiva. Observa que en este intervalo, la gráfica de $x^2$ está subiendo.
*   Para $x < 0$ (es decir, en el intervalo $(-\infty, 0)$), la derivada $f'(x) = 2x$ es negativa. Observa que en este intervalo, la gráfica de $x^2$ está bajando.
*   En $x = 0$, la derivada $f'(x) = 0$. Aquí es donde la función alcanza su mínimo, y la pendiente es horizontal.

### ¿Cuándo la Pendiente es Negativa?

La pendiente de una función en un punto dado (o el valor de su derivada en ese punto) se considera **negativa** cuando la función está **disminuyendo** o **decreciendo** en ese intervalo.

Visualmente, esto significa que si nos movemos de izquierda a derecha a lo largo del eje x:

*   La gráfica de la función **desciende**.
*   La línea tangente a la curva en ese punto se inclina **hacia abajo** de izquierda a derecha.

**En términos matemáticos:**

Si $f'(x) < 0$ (la derivada es menor que cero) en un intervalo $(a, b)$, entonces la función $f(x)$ es **decreciente** en ese intervalo. Esto implica que para cualquier par de puntos $x_1$ y $x_2$ en el intervalo, si $x_1 < x_2$, entonces $f(x_1) > f(x_2)$.

**Ejemplo:**

Retomando la función $f(x) = x^2$, su derivada es $f'(x) = 2x$.

*   Para $x < 0$ (es decir, en el intervalo $(-\infty, 0)$), la derivada $f'(x) = 2x$ es negativa (por ejemplo, si $x=-1$, $f'(-1) = -2$). En este intervalo, la gráfica de $x^2$ está descendiendo.
*   Para $x > 0$ (es decir, en el intervalo $(0, \infty)$), la derivada $f'(x) = 2x$ es positiva. La gráfica de $x^2$ está subiendo.
*   En $x = 0$, la derivada $f'(x) = 0$. Este es el punto donde la función cambia de decreciente a creciente, indicando un mínimo.

### La Dirección Negativa de la Derivada y la Búsqueda de Mínimos

Si la derivada de una función, $f'(x)$, es **negativa** en un punto dado, esto nos indica que la función $f(x)$ está **decreciendo** en ese punto. En otras palabras, a medida que aumentamos el valor de $x$, el valor de $f(x)$ disminuye.

Este concepto es fundamental en la optimización y en algoritmos como el descenso de gradiente (gradient descent), donde el objetivo es encontrar el mínimo de una función. Si estamos en un punto donde la derivada es negativa, moverse en la dirección positiva de $x$ nos llevaría a valores más bajos de la función, es decir, nos estaríamos "acercando" a un mínimo.

Considera la analogía de caminar por un terreno montañoso con los ojos vendados. Si estás en una ladera y sientes que la pendiente debajo de tus pies (la derivada) es negativa, sabes que si sigues caminando "cuesta abajo" (en la dirección donde $x$ aumenta si la derivada es negativa), llegarás a un punto más bajo. Continuar moviéndose en la dirección opuesta al gradiente (la dirección de la derivada negativa) es una estrategia clave para descender hacia un valle o un punto mínimo en el paisaje de la función.
