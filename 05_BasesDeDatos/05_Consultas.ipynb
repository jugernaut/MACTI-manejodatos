*Pandas* no es solo una módulo de *Python*; es el estándar de oro para la manipulación y el análisis de datos en *Python*. Se construye sobre *NumPy*, lo que significa que es rápida y eficiente pero añade una capa de abstracción necesaria para manejar datos estructurados/relacionados (tablas) de forma más accesible desde la perspectiva humana.

Los Pilares de *Pandas*

Para dominar esta herramienta, debes entender sus dos estructuras de datos fundamentales. Si no comprendes la diferencia, tu código será un complicado:

`Series`: Es una matriz unidimensional. Imagínala como una sola columna en una hoja de cálculo, capaz de contener cualquier tipo de dato (enteros, cadenas, flotantes, etc.).

`DataFrame`: Es la estructura principal, en español sería como "Marco de Datos". Una tabla bidimensional, con etiquetas en ejes (filas y columnas). Es, esencialmente, un diccionario de objetos Series.

¿Por qué es indispensable?

Un programador novato usa ciclos for para procesar datos; un experto usa *Pandas*. Esta biblioteca te permite:

Limpiar datos: Manejar valores nulos (`NaN`) que arruinarían cualquier análisis estadístico.

Alinear datos: Fusionar y unir conjuntos de datos distintos mediante claves comunes, tal como harías con un *JOIN* en *SQL*.

Transformar: Reestructurar tablas completas con una sola línea de código mediante el uso de *operaciones vectorizadas*.

Operaciones Vectorizadas: Las operaciones vectorizadas en *Pandas* permiten aplicar cálculos a columnas o series completas de datos de una sola vez, sin necesidad de utilizar bucles for explícitos. Esta técnica aprovecha la biblioteca *NumPy* subyacente para realizar operaciones optimizadas en *C/C++*, lo que resulta en un código mucho más rápido, conciso y fácil de leer.

Es posible crear diferentes `DataFrames`.

```python
datos_0 = {
    'Nombre' : ['Juan', 'Laura', 'Pepe'],
    'Edad': [42, 40, 37],
    'id_empleado': [1, 2, 3]
}

datos_1 ={
        'id_dpto': [4, 5, 6],
        'Departamento': ['Comunicación', 'Administración', 'Ventas'],
        'ventas': [1000, 5000, 5230]
}

datos_2 ={
        'id_empleado': [1, 2, 3],
        'id_dpto': [4, 5, 6],
}


vendedores = pd.DataFrame(datos_0)
ventas = pd.DataFrame(datos_1)
referencias = pd.DataFrame(datos_2)
```

Estos `DataFrames` se pueden unir (`JOIN`) mediante algún atributo, como se hace en *SQL*.

```python
q = '''
select *
from vendedores as v
join referencias as r
on v.id_empleado = r.id_empleado
'''
```

```python
sqldf(q)
```

Incluso podemos unir más de un `DataFrame`, recuerda que en *SQL* toda consulta devuelve una tabla (`DataFrame`), misma a la cuál se pueden aplicar una nueva consulta.

```python
q = '''
select *
from vendedores as v
join referencias as r
on v.id_empleado = r.id_empleado
join ventas as vt
on vt.id_dpto = r.id_dpto
'''
```

```python
sqldf(q)
```

## Consumir datos desde un .csv

Otra cosa muy valiosa es consumir datos desde un .csv

```python
import pandas as pd
from pandasql import sqldf

# 1. Carga de datos
alumnos = pd.read_csv('alumnos.csv')
grupos = pd.read_csv('grupos.csv')

# 2. Validación estricta de tipos
if alumnos['id_grupo'].dtype != grupos['id_grupo'].dtype:
    print("ADVERTENCIA: Tipos de datos incompatibles detectados. Corrigiendo...")
    alumnos['id_grupo'] = alumnos['id_grupo'].astype(int)
    grupos['id_grupo'] = grupos['id_grupo'].astype(int)

# 3. Consulta Relacional
query = """
    SELECT
        a.nombre AS alumno,
        g.nombre_grupo AS curso,
        g.salon
    FROM
        alumnos a
    INNER JOIN
        grupos g ON a.id_grupo = g.id_grupo
"""

# 4. Ejecución
resultado = sqldf(query, locals())
print(resultado)
```

# Conexión a una Base de Datos

Ya que vimos como crear una base de datos en *Pandas*, ahora vamos a extraer y cargar los datos de una base de datos en *Pandas*. Esto tiene 2 principales ventajas:

* Los datos están en memoria, es decir que no es necesario abrir una conexión y conectarse a la base de datos, cada vez que necesitemos realizar alguna consulta.

* Las consultas se realizan mediante *SQL* pero mediante la *Pandas*.

Para conectar con *MySQL* necesitarás instalar `pymysql` o `mysql-connector-python` para que `SQLAlchemy` pueda hablar con el servidor.

Una vez instaladas las bibliotecas necesarias hay que establecer la conexión a la base de datos y realizar consultas.

A continuación se muestra como realizar la consulta mediante `mysql-connector-python`.

```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.exc import SQLAlchemyError

# Configuración técnica
CONFIG = {
    'user': 'manejo',
    'pass': 'prometeo',
    'host': 'hpc-matematicas-z.fciencias.unam.mx',
    'port': '3306',
    'db':   'FCiencias'
}

# 1. Construcción de la URL del motor
# Usamos f-strings para mantener el código limpio y legible
url_motor = f"mysql+mysqlconnector://{CONFIG['user']}:{CONFIG['pass']}@{CONFIG['host']}:{CONFIG['port']}/{CONFIG['db']}"

try:
    # 2. Creación del motor
    engine = create_engine(url_motor)

    # 3. Extracción con contexto seguro
    # El bloque 'with' garantiza que la conexión se libere al terminar
    with engine.connect() as conn:
        print("Conexión establecida con mysql-connector-python...")

        query = "SELECT * FROM ALUMNOS"
        df = pd.read_sql(query, conn)

    # 4. Verificación de integridad
    if df.empty:
        print("Advertencia: La consulta no devolvió registros.")
    else:
        print(f"Éxito: Se han cargado {df.shape[0]} filas y {df.shape[1]} columnas.")
        print(df.head()) # Muestra las primeras 5 filas para validar

except SQLAlchemyError as e:
    print(f"Error de SQLAlchemy: {e}")
except Exception as e:
    print(f"Error inesperado: {e}")
```

Ya con los datos almacenados en `DataFrames`, podemos consultarlos mediante *Pandas*

```python
from ipywidgets import interact, interactive, fixed
import ipywidgets as widgets
import numpy as np
from pandasql import sqldf


def consulta_alumnos():
  consulta = '''SELECT nombre_tutor FROM tutores;'''
  vista = sqldf(consulta)
  nombre_tutores = vista['nombre_tutor'].tolist()
  return nombre_tutores

def consulta_tutores(tutor):
  consulta = f"""SELECT nombre FROM alumnos NATURAL JOIN tutorias NATURAL JOIN
             tutores WHERE nombre_tutor = '{tutor}'"""
  vista = sqldf(consulta)
  nombre_alumnos = vista['nombre'].tolist()
  print(nombre_alumnos)



w = interact(consulta_tutores, tutor = consulta_alumnos())
```
