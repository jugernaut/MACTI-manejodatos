## 🌱 Parte 1: ¿Qué son las Expresiones Regulares?


### Definición

Una **expresión regular** (también llamada *regex*) es una secuencia de caracteres que define un **patrón de búsqueda**. Se usan en:

- 📊 **Análisis de datos**: limpieza y extracción de información
- 🌐 **Desarrollo web**: validación de correos y contraseñas
- 🔒 Ciberseguridad: detección de patrones maliciosos en logs y análisis de tráfico de red
- 📜 **Procesamiento de lenguaje natural**: tokenización y limpieza de texto antes de entrenar modelos
- 🧬 **Bioinformática**: búsqueda de motivos en DNA y proteínas
- 💊 **Medicina**: minería de texto en registros clínicos


### Una analogía biológica

Imagina que eres investigadora en un laboratorio de genómica. Tienes una secuencia de DNA de 10,000 nucleótidos y necesitas encontrar **todos los codones de inicio** (`ATG`).

Podrías:
- ❌ Leerla letra por letra → *imposible*
- ❌ Escribir código que compare cada posición → *tedioso*
- ✅ Usar una **expresión regular** → *una línea de código, resultado instantáneo*


## 🐍 Parte 2: Regex en Python — el módulo `re`

Python incluye el módulo `re` en su biblioteca estándar. No necesitas instalar nada.

| Función | ¿Qué hace? | Devuelve |
|---------|-----------|----------|
| `re.search(patrón, texto)` | Primera coincidencia en cualquier parte | Objeto Match o None |
| `re.match(patrón, texto)` | Coincidencia solo al inicio | Objeto Match o None |
| `re.findall(patrón, texto)` | Todas las coincidencias | Lista de strings |
| `re.finditer(patrón, texto)` | Todas las coincidencias con posición | Iterador de Match |
| `re.sub(patrón, reemplazo, texto)` | Reemplaza coincidencias | String modificado |

### Empecemos viendo algunos ejemplos

### 2.1 `re.search()`

La función `re.search()` recorre toda la cadena buscando la primera vez que aparece el patrón. Si lo encuentra, devuelve un objeto con información sobre la coincidencia (posición, texto encontrado). Si no lo encuentra, devuelve None. Es útil cuando solo necesitas saber si el patrón existe y dónde aparece por primera vez.

```python
import re

# Ejemplo: ¿está el codón ATG en esta secuencia?
secuencia = "GCTATGCGATTAGCTA"
patron = "ATG"

resultado = re.search(patron, secuencia)

if resultado:
    print(f"Patrón encontrado en la posición {resultado.start()}")
else:
    print("Patrón no encontrado")
```

### 2.2 `re.findall()`

A diferencia de `search`, `re.findall()` no se detiene en la primera coincidencia: devuelve una lista con todas las apariciones del patrón en el texto. Si no hay ninguna, devuelve una lista vacía. Es la función más usada cuando necesitas contar o recolectar todas las ocurrencias.

```python
# findall: encontrar TODOS los codones de inicio
secuencia_larga = "ATGCGATTAGCTATGCCCATGTAGGCTA"

codones = re.findall("ATG", secuencia_larga)
print(f"Codones ATG encontrados: {codones}")
print(f"Total: {len(codones)}")
```

### 2.3 `re.finditer`

`re.finditer()` hace lo mismo que findall pero en lugar de devolver solo el texto encontrado, devuelve un iterador de objetos Match. Cada objeto contiene no solo la cadena coincidente sino también su posición exacta. Es preferible a findall cuando necesitas saber dónde está cada coincidencia, no solo cuál es.

```python
# finditer: encontrar posiciones exactas
secuencia_larga = "ATGCGATTAGCTATGCCCATGTAGGCTA"

for match in re.finditer("ATG", secuencia_larga):
    print(f"ATG encontrado en posición {match.start()} — {match.end()}")
```

## 🔣 Parte 3: Metacaracteres — el verdadero poder

Hasta ahora hemos buscado texto literal, es decir, exactamente lo que escribimos. Los metacaracteres son símbolos con un significado especial dentro de una expresión regular: no representan el carácter en sí mismo, sino una instrucción sobre cómo buscar. Por ejemplo, el punto . no busca un punto literal, sino que significa "cualquier carácter". Combinando metacaracteres podemos describir patrones flexibles y poderosos en una sola expresión.

| Símbolo | Significado | Ejemplo | Coincide con |
|---------|-------------|---------|-------------|
| `.` | Cualquier carácter | `A.G` | ACG, ATG, AGG... |
| `*` | 0 o más del anterior | `AT*G` | AG, ATG, ATTG... |
| `+` | 1 o más del anterior | `AT+G` | ATG, ATTG... (no AG) |
| `?` | 0 o 1 del anterior | `AT?G` | AG o ATG |
| `{n}` | Exactamente n | `A{3}` | AAA |
| `{n,m}` | Entre n y m | `A{2,3}` | AA o AAA |
| `[abc]` | a, b o c | `[ATG]` | A, T o G |
| `[^abc]` | Ninguno de ellos | `[^ATG]` | C u otra letra |
| `^` | Inicio del texto | `^ATG` | Solo si empieza con ATG |
| `$` | Final del texto | `TAA$` | Solo si termina con TAA |
| `\|` | O (alternativa) | `ATG\|TTG` | ATG o TTG |
| `()` | Grupo | `(AT)+` | AT, ATAT... |

### Ejemplos prácticos. 

1. Imagina que trabajas en un laboratorio de genética forense. Te piden identificar microsatélites: pequeñas regiones del DNA donde un patrón (como CA) se repite consecutivamente. Estas repeticiones varían entre individuos, por lo que se usan como "huella genética" para identificar personas o establecer parentescos. En el siguiente ejemplo buscamos repeticiones de CA entre 3 y 6 veces, y verás la posición exacta donde aparece cada una.

```python
# Secuencia de DNA donde realizaremos la búsqueda
secuencia_genomica = "ATGCACACACACATGCAGCAGCATTT"

# Patrón: el dinucleótido CA repetido entre 3 y 6 veces consecutivas
# (CA){3,6} significa: el grupo "CA" debe aparecer mínimo 3 y máximo 6 veces seguidas
patron_microsatelite = "(CA){3,6}"

# finditer nos da cada coincidencia con su posición exacta en la secuencia
for match in re.finditer(patron_microsatelite, secuencia_genomica):
    # match.group() → la subcadena que coincidió con el patrón
    # match.start() → la posición donde empieza (contando desde 0)
    print(f"Microsatélite: '{match.group()}' en posición {match.start()}")
```

2. Imagina que trabajas analizando proteínas de membrana. Necesitas identificar dominios transmembrana: regiones de la proteína suficientemente hidrofóbicas como para atravesar la membrana celular. Estas regiones están formadas por aminoácidos que "repelen" el agua: A, V, I, L, M, F, W y P. Usando una clase de caracteres [AVILMFWP] podemos buscar tramos donde estos aminoácidos aparezcan consecutivamente. Al ejecutar el código verás qué regiones de la proteína son candidatas a ser dominios transmembrana.

```python
# Clases de caracteres en proteínas: buscar regiones hidrofóbicas

# Secuencia de proteína en formato de una letra por aminoácido
secuencia_proteina = "MKTAYIAKQRQISFVKSHFSRQLEERLGLIEVQAPILSRVGDGTQDNLSGAEKAVQVKVKALPDAQNTASSGGSSGA"

# Patrón: cualquier aminoácido hidrofóbico [AVILMFWP], 4 o más veces consecutivas
# Los corchetes [] definen un conjunto: "cualquiera de estos caracteres"
# {4,} significa: al menos 4 veces seguidas (sin límite superior)
patron_hidrofobico = "[AVILMFWP]{4,}"

print("Regiones hidrofóbicas encontradas:")
for match in re.finditer(patron_hidrofobico, secuencia_proteina):
    # match.start()+1 → convertimos de índice base 0 a base 1 (más intuitivo en biología)
    print(f"  '{match.group()}' — posición {match.start()+1} a {match.end()}")
```

3. Antes de analizar una secuencia, es fundamental verificar que los datos sean válidos. Imagina que recibes secuencias de DNA de distintas fuentes: algunas pueden contener caracteres inválidos (espacios, letras que no son nucleótidos, saltos de línea) producto de errores de formato. Las anclas ^ y $ nos permiten exigir que el patrón cubra la cadena completa, de inicio a fin, no solo una parte. Al ejecutar el código verás cuáles secuencias pasan el control de calidad y cuáles son rechazadas.


```python
# Anclas ^ y $: validar que una secuencia de DNA sea completamente válida

secuencias_test = [
    "ATGCGATCG",      # válida: solo contiene A, T, C, G
    "ATGCX ATCG",     # inválida: contiene X y un espacio
    "atgcgatcg",      # inválida: minúsculas no permitidas
    "ATGCGATCG",      # válida
]

# ^ exige que el patrón empiece desde el inicio de la cadena
# $ exige que el patrón llegue hasta el final de la cadena
# [ATCG]+ significa: uno o más caracteres que sean A, T, C o G
# Combinados: toda la cadena, de principio a fin, debe ser solo A, T, C o G
patron_valido = "^[ATCG]+$"

for seq in secuencias_test:
    # re.match busca el patrón desde el inicio de la cadena
    if re.match(patron_valido, seq):
        print(f"✅ '{seq}' — secuencia válida")
    else:
        print(f"❌ '{seq}' — secuencia inválida")
```

## 🔬 Parte 4: Caso Práctico — Marco de Lectura Abierto (ORF)

Un **ORF (Open Reading Frame)** es una región de DNA que:
- Comienza con el codón de inicio: `ATG`
- Termina con un codón de paro: `TAA`, `TAG` o `TGA`

Identificar ORFs es el primer paso para encontrar genes.

```python
# ORF completo: ATG + nucleótidos intermedios + codón de paro
# El ? hace la búsqueda NO codiciosa (busca el ORF más corto)

secuencia = "GCTATGAAACCCGGGTTTAAAGCTA" # Secuencia de DNA que contiene un ORF embebido
patron_orf = "ATG[ATCG]+?(TAA|TAG|TGA)"

# re.search devuelve solo la primera coincidencia
match = re.search(patron_orf, secuencia)
if match:
    print(f"ORF encontrado     : {match.group()}") # secuencia completa del ORF
    print(f"Codón de paro      : {match.group(1)}")  # group(1) → contenido del primer grupo ()
    print(f"Posición           : {match.start()} — {match.end()}") # índices base 0
    print(f"Longitud           : {len(match.group())} nucleótidos")
```
