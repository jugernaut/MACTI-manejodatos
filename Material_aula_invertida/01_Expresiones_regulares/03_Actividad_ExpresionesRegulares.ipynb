## 🟢 Nivel 1 — Calentamiento (5 min)

### Tarea 1: Control de Calidad de Secuencias

**Correo recibido:**
> *"El secuenciador tuvo un problema. Algunas secuencias salieron con caracteres extraños. Filtren las secuencias válidas (solo A, T, C, G) y reporten cuántas son válidas y cuántas no."*

```python
# Datos de entrada — NO modificar
secuencias_brutas = [
    "ATGCGATCGCATGC",
    "ATGCN ATCG",
    "GGCTATCGATCGAT",
    "ATGX123CGATCG",
    "CGATCGATCGATCG",
    "atgcgatcg",
    "TTTTAAAACCCCGGGG",
    "ATG-CGA",
    "GCATGCATGCATGC"
]
print(f"Total de secuencias recibidas: {len(secuencias_brutas)}")
```

```python
# ✏️ SU CÓDIGO AQUÍ
# Pista: usen el patrón ^[ATCG]+$ del material de pre-clase

validas = []
invalidas = []

patron = None  # <- completen este patrón

for seq in secuencias_brutas:
    if re.match(patron, seq):  # <- ¿qué función de re usan aquí?
        validas.append(seq)
    else:
        invalidas.append(seq)

print(f"Secuencias VÁLIDAS ({len(validas)}):")
for s in validas:
    print(f"  ✅ {s}")

print(f"\nSecuencias INVÁLIDAS ({len(invalidas)}):")
for s in invalidas:
    print(f"  ❌ {repr(s)}")
```

## 🟡 Nivel 2 — El desafío principal (15 min)

### Tarea 2: Búsqueda de Genes Candidatos

**Correo recibido:**
> *"Tenemos un fragmento genómico. Identifiquen todos los posibles genes: (1) empiezan con ATG, (2) terminan con TAA, TAG o TGA, (3) tienen longitud múltiplo de 3. Reporten posición, longitud y número de codones."*

```python
# Genoma de trabajo — NO modificar
GENOMA = (
    "TATAAAAATGCCCGGGTTTTAAATGCGATCGCGATGA"
    "GCTATGAAACCCGGGTTTTAAGCTATGCAGTAGTAA"
    "ATGAAACGATGACCCATGCCCATGTAG"
)
print(f"Longitud del genoma: {len(GENOMA)} nucleótidos")
print(GENOMA)
```

```python
# ✏️ SU CÓDIGO AQUÍ

# Paso 1: definan el patrón para ORF (no codicioso)
patron_orf = None  # <- completen aquí

genes_candidatos = []

for match in re.finditer(patron_orf, GENOMA):
    secuencia_gen = match.group()
    longitud = len(secuencia_gen)

    # Paso 2: filtrar solo múltiplos de 3
    if longitud % 3 == 0:
        genes_candidatos.append({
            'secuencia': secuencia_gen,
            'inicio': match.start(),
            'fin': match.end(),
            'longitud': longitud,
            'num_codones': longitud // 3
        })

# Paso 3: reporte
print(f"Genes candidatos encontrados: {len(genes_candidatos)}\n")
for i, gen in enumerate(genes_candidatos, 1):
    print(f"Gen #{i}")
    print(f"  Secuencia : {gen['secuencia']}")
    print(f"  Posición  : {gen['inicio']} — {gen['fin']}")
    print(f"  Longitud  : {gen['longitud']} nt | {gen['num_codones']} codones")
    print()
```

```python
# BONUS: estadísticas del genoma
if genes_candidatos:
    gen_largo = max(genes_candidatos, key=lambda g: g['longitud'])
    gen_corto = min(genes_candidatos, key=lambda g: g['longitud'])
    nt_cod = sum(g['longitud'] for g in genes_candidatos)
    pct = nt_cod / len(GENOMA) * 100

    print(f"Gen más largo  : {gen_largo['secuencia']} ({gen_largo['longitud']} nt)")
    print(f"Gen más corto  : {gen_corto['secuencia']} ({gen_corto['longitud']} nt)")
    print(f"% codificante  : {pct:.1f}%")
```

## 🔴 Nivel 3 — El reto final (10 min)

### Tarea 3: Análisis de Proteínas — elijan al menos 2 misiones

```python
# Proteínas de trabajo — NO modificar
PROTEINAS = {
    "ProtA": "MKKRGDSTVNITSIKDELHHHHPVIAFWLLATNLSKDEL",
    "ProtB": "MPFLLLSAVVLLLNASTKRGDNASTNITSRRGDRGD",
    "ProtC": "MAAVILMFWPLLSRGDNITSIKKDELKDEL"
}
for nombre, seq in PROTEINAS.items():
    print(f"{nombre} ({len(seq)} aa): {seq}")
```

```python
# ─── MISIÓN A: Buscar motivos funcionales ───
# Completen los patrones que faltan

motivos = {
    "Adhesion (RGD)"    : "RGD",
    "N-glicosilacion"   : None,   # <- patrón N[^P][ST]
    "Retencion RE (KDEL)": None,  # <- KDEL al final
    "NLS (basica)"      : None    # <- 2+ residuos K o R
}

print("=" * 55)
print("REPORTE DE MOTIVOS FUNCIONALES")
print("=" * 55)

for nombre_prot, secuencia in PROTEINAS.items():
    print(f"\n{nombre_prot}:")
    for nombre_motivo, patron in motivos.items():
        if patron:
            encontrados = re.findall(patron, secuencia)
            if encontrados:
                print(f"  ✅ {nombre_motivo}: {encontrados}")
            else:
                print(f"  ❌ {nombre_motivo}: no encontrado")
        else:
            print(f"  ⏳ {nombre_motivo}: patrón pendiente")
```

```python
# ─── MISIÓN B: Composición de aminoácidos ───
grupos = {
    "Hidrofobicos" : "AVILMFWP",
    "Polares"      : "STCYNQ",
    "Cargados (+)" : "KRH",
    "Cargados (-)" : "DE",
    "Otros (Gly)"  : "G"
}

print("=" * 55)
print("COMPOSICION DE AMINOACIDOS")
print("=" * 55)

for nombre_prot, secuencia in PROTEINAS.items():
    total = len(secuencia)
    print(f"\n{nombre_prot} ({total} aa):")
    for grupo, aas in grupos.items():
        encontrados = re.findall(f"[{aas}]", secuencia)
        pct = len(encontrados) / total * 100
        barra = "█" * int(pct / 5)
        print(f"  {grupo:15}: {len(encontrados):3} aa ({pct:5.1f}%) {barra}")
```
